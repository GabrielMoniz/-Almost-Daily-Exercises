Wigner function for for a squeezed thermal state is $$W(q,p)=\frac{\exp\!\left(-\frac{q^2}{2V_q}-\frac{p^2}{2V_p}\right)}{2\pi\sqrt{V_qV_p}}.$$
where $V_q=(2n+1)e^{-2r}/2$ and $V_p=(2n+1)e^{2r}/2$. Here $r < 0$ squeezes momentum and $n$ is the mean number of thermal photons.

The code define the functions to calculate the wigner function on a grid and plot it.

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.cm as cm

def squeezed(r, n):

  V_q = (2*n+1)*np.exp(-2*r)/2
  V_p = (2*n+1)*np.exp(2*r)/2

  return V_q, V_p

def gauss_wigner(q, p, V_q, V_p):
  denominator = 2*np.pi*np.sqrt(V_q*V_p)
  W = np.exp(-q**2/(2*V_q) - p**2/(2*V_p))/denominator

  return W



#generating phase lattice
x = np.linspace(-5, 5, 401)
y = np.linspace(-5, 5, 401)

q, p = np.meshgrid(x, y, indexing="ij")



cases = [("vacuum", 0, 0), ("squeezed vacuum", 0, np.log(2)/2), ("two photon thermal state", 2, 0), ("squeezed two photon thermal state", 2, np.log(2)/2)]   # a list of tuples
for label, n, r in cases:
  #calculating wigner function for thermal state

  V_q , V_p = squeezed(r, n)
  print("V_q, V_p:", V_q, V_p)
  print("W(0,0):", gauss_wigner(0, 0, V_q, V_p) )

  W = gauss_wigner(q, p, V_q, V_p)



  #plotting phase lattice
  plt.figure()
  plt.title(f'{label} n={n} r={r:.2f}')
  plt.contourf(q, p, W, levels=50); plt.colorbar()
  plt.xlabel("q"); plt.ylabel("p"); plt.gca().set_aspect("equal")
  plt.show()                     # each tuple is unpacked into three names

```

Wigner Function calculator for one-mode for a pure state. The integral $$W_\phi(q,p):=\frac{1}{\pi}\int dy\;e^{2ip y}\,
  \phi(q-y)\phi^*(q+y),$$ can be approximated by a sum over y

  $$W_\phi(q,p) \approx \frac{dy}{\pi} \sum_i e^{2ip y_i}\,
  \phi(q-y_i)\phi^*(q+y_i)$$
  
For plotting on a grid, we can define a matrix $W_{qy}$ such that

$$W_{qp} = \frac{dy}{\pi} \sum_y G_{qy}E_{yp} \iff W = \frac{dy}{\pi} G E$$

generating the whole grid, where
$$G_{qy} = \phi(q-y)\phi^*(q+y)\quad E_{yp}=e^{2iyp} $$

```python
#calculates the Wigner function for a general one mode state
def Wigner_function(q, p, phi):
  dy = 0.01
  y = np.arange(-10, 10 + dy, dy)
  # matrix E_{yp}
  E = np.exp(2j*np.outer(y, p))
  # matrix G_{qy}
  G = phi(q[ :, None] - y[None, :]) * np.conj(phi(q[:, None] + y[None, :]))
  # matrix \sum_y E_pyG_yq
  integrand = G @ E

  return (dy*integrand/np.pi).real


#generating phase lattice
q = np.linspace(-7, 7, 500)
p = np.linspace(-7, 7, 500)


# the wave function
phi = lambda x: np.pi**-0.25 * np.exp(-x**2 / 2)




cases = [
    {"gamma": 0.0, "q_range": (-4, 4), "p_range": (-4, 4)},
    {"gamma": 0.1, "q_range": (-4, 4), "p_range": (-4, 4)},
    {"gamma": 0.5, "q_range": (-4, 4), "p_range": (-4, 12)},
]
for case in cases:
  gamma = case["gamma"]
  q = np.linspace(*case["q_range"], 500)
  p = np.linspace(*case["p_range"], 500)

  psi = lambda x: phi(x) * np.exp(1j * gamma * x**3)

  W = Wigner_function(q, p, psi)

  #plotting phase lattice
  plt.figure()
  plt.title(f'Wigner function for $\gamma=${gamma}')
  plt.contourf(q, p, W.T, levels=50); plt.colorbar()
  plt.xlabel("q"); plt.ylabel("p"); plt.gca().set_aspect("equal")
  plt.show()                     # each tuple is unpacked into three names
```

Modelling gaussian noise

It acts as $$W_{\rm out}(q,p)=\frac1T\,\big[W*N_{v/T}\big]\!\left(\frac{q}{\sqrt T},\frac{p}{\sqrt T}\right),$$ for $$
  N_v(x):=\frac{1}{\sqrt{2\pi v}}\,e^{-x^{2}/(2v)},\qquad x\in\mathbb R ^2$$

```python
from scipy.ndimage import gaussian_filter
# calculates the wigner function for the output state after the initial state suffers gaussian noise
# variance v and transmissivity T
def Wigner_convolution(q, p, psi , v, T):

  dq, dp = q[1] - q[0], p[1] - p[0]

  W_in = Wigner_function(q/np.sqrt(T), p/np.sqrt(T), psi)

  W_out = gaussian_filter(W_in, sigma=(np.sqrt(v)/dq, np.sqrt(v)/dp), mode="constant", truncate=8.0)/T

  return W_out


def matrix_integral(W, dp, dq):
  # W: array with one axis per phase-space coordinate, on a uniform grid along each axis
  # dq, dp: the spacing of each q-axis and each p-axis (numbers or arrays)
  S = np.sum(W)
  return S*np.prod(dp)*np.prod(dq)

def Wigner_negativity(W, dp, dq):

  Normalization = matrix_integral(W, dp, dq)

  S_abs = np.sum(np.abs(W))
  integral = S_abs*np.prod(dp)*np.prod(dq)

  return  (integral - Normalization)/2
```

Here we will compare the wigner function for the input and the output state

```python
from matplotlib.colors import TwoSlopeNorm


cases_kick = [
    {"gamma": 0.0, "q_range": (-4, 4), "p_range": (-4, 4), "v": 0.1, "T":1},
    {"gamma": 0.1, "q_range": (-4, 4), "p_range": (-4, 4), "v": 0.1, "T":1},
    {"gamma": 0.5, "q_range": (-4, 4), "p_range": (-4, 12), "v": 0.1, "T":1},
]
for case in cases_kick:
  gamma = case["gamma"]
  q = np.linspace(*case["q_range"], 500)
  p = np.linspace(*case["p_range"], 500)

  dq = q[1] - q[0]
  dp = p[1] - p[0]

  psi = lambda x: phi(x) * np.exp(1j * gamma * x**3)

  v = case["v"]
  T = case["T"]

  W_out = Wigner_convolution(q, p, psi , v, T)
  W_in = Wigner_function(q, p, psi)

  print("Normalization check W_in ", round(matrix_integral(W_in, dp, dq),4) )
  print("Normalization check W_out ", round(matrix_integral(W_out, dp, dq),4) )
  print("Negativity W_in ", round( Wigner_negativity(W_in, dp, dq),4) )
  print("Negativity W_out ", round(Wigner_negativity(W_out, dp, dq),4) )

  lo = min(W_out.min(), W_in.min(), -1e-12)      # most negative value of BOTH panels (kept < 0)
  hi = max(W_out.max(), W_in.max())              # largest value of BOTH panels

  norm = TwoSlopeNorm(vmin=lo, vcenter=0.0, vmax=hi)

  #plotting phase lattice side by side
  fig, axes = plt.subplots(1, 2, figsize=(10, 4), sharex=True, sharey=True, constrained_layout=True)
  fig.suptitle(f'Wigner function for $\\gamma={gamma}$', fontsize=14)

  # Plot input Wigner function on the first axis (axes[0])
  im0 = axes[0].pcolormesh(q, p, W_in.T, cmap="RdBu_r", norm=norm, shading="auto")
  axes[0].set_title("Input $W_{in}$ (Pure State)")
  axes[0].set_xlabel("q")
  axes[0].set_ylabel("p")
  axes[0].set_aspect("equal")

  # Plot output Wigner function on the second axis (axes[1])
  im1 = axes[1].pcolormesh(q, p, W_out.T, cmap="RdBu_r", norm=norm, shading="auto")
  axes[1].set_title(f"Output $W_{{out}}$ (Noise v={v}, T={T})")
  axes[1].set_xlabel("q")
  axes[1].set_aspect("equal")

  # Add a single colorbar associated with the normalization
  fig.colorbar(im1, ax=axes, shrink=0.8)
  plt.show()
```

# Fock states

The output state of fock state $|n\rangle$ after a pure loss with transmissivity T is
 $$ \rho_T = \sum_{k=0}^{n}  \binom{n}{k}T^{k} (1-T)^{n-k} |k\rangle\langle k| $$
which has Wigner function given by

$$  W_{\rho_T} (\alpha)= \frac{e^{-2|\alpha|^2}}{\pi} (1-2T)^n  \mathbf{L}_n\left(\frac{4T}{2T-1} |\alpha|^2\right); \quad \alpha = \frac{x+ip}{\sqrt{2}}$$

Where $\mathbf L_n$ denotes the n-th laguerre polynomial.

Notice to integrate over the phase-space, we can make the substituion of variables $s=|\alpha|^2 = (q^2 + p^2)/2$, and due to the rotational symmetry $$\int dqdp = 2\pi \int ds$$

```python
from scipy.special import eval_laguerre
from scipy.integrate import trapezoid
import ipywidgets as widgets
from ipywidgets import interact
from matplotlib.colors import TwoSlopeNorm


def Transmited_fock(q, p, n, T):
  factor = np.exp(-(q**2 +p**2))*(1-2*T)**n/np.pi

  argument = (q**2 +p**2)*2*T/(2*T-1)
  L_n = eval_laguerre(n, argument)

  return factor*L_n


def Transmitted_fock_negativity(n,T):
  s= np.linspace(0, 10, 500)
  ds = s[1] - s[0]
  factor = 2*np.exp(-2*s)*(1-2*T)**n

  argument = s*4*T/(2*T-1)
  L_n = eval_laguerre(n, argument)

  W_s = factor*L_n

  return trapezoid(np.maximum(-W_s, 0), s)



L = 4
x = np.linspace(-L, L, 400)
q, p = np.meshgrid(x, x, indexing="ij")          # built once, reused by every redraw

def show(n=1, T=0.8):
    if abs(2*T - 1) < 1e-9:                      # the formula divides by (2T-1); see below
        T = 0.5 + 1e-9
    W = Transmited_fock(q, p, n, T)
    delta = Transmitted_fock_negativity(n, T)

    norm = TwoSlopeNorm(vmin=min(W.min(), -1e-12), vcenter=0.0, vmax=max(W.max(), 1e-12))
    fig, ax = plt.subplots(figsize=(5, 4.5))
    im = ax.pcolormesh(x, x, W.T, cmap="RdBu_r", norm=norm, shading="auto")
    fig.colorbar(im, ax=ax)
    ax.set_title(rf"$n={n}$, $T={T:.2f}$, $v_n={delta:.4f}$")
    ax.set_xlabel("q"); ax.set_ylabel("p"); ax.set_aspect("equal")
    plt.show()

interact(show,
         n=widgets.IntSlider(value=1, min=1, max=12, description="n"),
         T=widgets.FloatSlider(value=0.8, min=0.0, max=1.0, step=0.01,
                               description="T", continuous_update=False))
```

# Multimode states

To analyze n+1-mode states, we can use the interact plotting the surface level for the first mode (plotting in variables $q_1$ and $p_1$), while leaving the other phase-variables $q_i$, $p_i$, for $i>1$ as constants, that can ajusted parameters using interact again. For non-entangled states, varying the phase of other modes only reescales the restricted Wigner surface for the first mode.
Trying to find the entire grid is undoable due to lack of memory.



To adjust the Wigner grid calculater, we must notice that for a $k$-mode state ($\mathbf{y} \in \mathbb{R}^k$) our grid approximation is
$$W_{\mathbf{q}\,\mathbf{p}} = \frac{dy}{\pi} \sum_{\mathbf{y}} G_{\mathbf{q}\, \mathbf{y}}E_{\mathbf{y}\,\mathbf{p}} $$ being $G,E$ tensors of rank $2k$

  
$$G_{\mathbf{q}\, \mathbf{y}} = \psi(\mathbf{q}-\mathbf{y})\phi^*(\mathbf{q}+ \mathbf{y})\quad E_{yp}=e^{2iyp} $$

```python
from IPython.display import Image; Image("tensor_network.png")
```

```python
#calculates the Wigner function for a general one mode state
def Wigner_multimode(q1, p1, psi, fixed, y):
    # fixed = [(q2, p2), (q3, p3), ...];  k = 1 + len(fixed)
    # returns W on the (q1, p1) grid, indexed [q1, p1]
    k = 1 + len(fixed)
    Y = np.meshgrid(*([y] * k), indexing="ij")                       # all k-tuples (y_1, ..., y_k)
    phase = np.prod([np.exp(2j * pm * Y[m + 1]) for m, (_, pm) in enumerate(fixed)], axis=0)
    E1 = np.exp(2j * np.outer(p1, y))                                # E1[b, j] = exp(2i p1_b y_j)
    W = np.empty((q1.size, p1.size))
    for i, a in enumerate(q1):
        x = [a] + [qm for qm, _ in fixed]                            # the point (q1, q2, ..., qk)
        G = psi(*[x[m] - Y[m] for m in range(k)]) * np.conj(psi(*[x[m] + Y[m] for m in range(k)]))
        g1 = (G * phase).reshape(y.size, -1).sum(axis=1)             # sum over y_2..y_k
        W[i] = (E1 @ g1).real
    return W * (y[1] - y[0])**k / np.pi**k
```

Take the state
$|\psi\rangle=e^{i\gamma\hat q_1\hat q_2}|r\rangle|r\rangle$, with wavefunction
$$\psi(x_1,x_2)=\varphi_r(x_1)\,\varphi_r(x_2)\,e^{i\gamma x_1x_2},\qquad \varphi_r(x)=(2\pi V_q)^{-1/4}e^{-x^2/(4V_q)}.$$

```python
Vq = 0.5
gamma = 1
# the wave function
phi_r = lambda x: (2*Vq*np.pi)**(-0.25) * np.exp(-x**2 / (4*Vq))

psi2 = lambda x1, x2: phi_r(x1) * phi_r(x2) * np.exp(1j * gamma * x1 * x2)



L = 4
x = np.linspace(-2.5, 2.5, 200) # built once both q_1 and p_1 grid

def show(q_2=1, p_2=1):

    y = np.linspace(-6, 6, 200)

    fixed= [(q_2, p_2)]

    W = Wigner_multimode(x, x, psi2, fixed, y)


    norm = TwoSlopeNorm(vmin=min(W.min(), -1e-12), vcenter=0.0, vmax=max(W.max(), 1e-12))
    fig, ax = plt.subplots(figsize=(5, 4.5))
    im = ax.pcolormesh(x, x, W.T, cmap="RdBu_r", norm=norm, shading="auto")
    fig.colorbar(im, ax=ax)
    ax.set_title(rf"$q_2={q_2:.2f}$, $p_2={p_2:.2f}$")
    ax.set_xlabel("q"); ax.set_ylabel("p"); ax.set_aspect("equal")
    plt.show()

interact(show,
         q_2=widgets.FloatSlider(value=0.8, min=0.0, max=1.0, step=0.1,
                               description="q_2", continuous_update=False),
         p_2=widgets.FloatSlider(value=0.8, min=0.0, max=1.0, step=0.1,
                               description="p_2", continuous_update=False))
```
